# Temperature Prediction - EDA

Exploratory data analysis for weather data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
```

```python
# Load data
df = pd.read_csv('../data/weather.csv')
df['date'] = pd.to_datetime(df['date'])
df.head()
```

```python
# Basic stats
print(f"Date range: {df['date'].min()} to {df['date'].max()}")
print(f"Total days: {len(df)}")
df.describe()
```

```python
# Max temperature distribution
fig, ax = plt.subplots(figsize=(12, 4))
df['temp_max_real'].hist(bins=50, ax=ax)
ax.set_xlabel('Max Temperature (°C)')
ax.set_ylabel('Frequency')
ax.set_title('Distribution of Daily Max Temperature')
plt.show()
```

```python
# Monthly max temp
df['month'] = df['date'].dt.month
monthly = df.groupby('month')['temp_max_real'].agg(['mean', 'std', 'min', 'max'])
print(monthly)
```

```python
# Correlation matrix
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr = df[numeric_cols].corr()

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, ax=ax)
plt.title('Feature Correlation Matrix')
plt.show()
```

```python
# Scatter: morning temp vs max temp
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(df['temp_06h'], df['temp_max_real'], alpha=0.3, s=5)
ax.set_xlabel('Morning Temperature (06h)')
ax.set_ylabel('Max Temperature')
ax.set_title('Morning Temp vs Max Temp')
plt.show()
```
