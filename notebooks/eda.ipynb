# Temperature Prediction - EDA

Exploratory data analysis for rp5.lv SBGR weather data (2012-2026).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
%matplotlib inline
```

```python
# Load daily processed data
df = pd.read_csv('/tmp/weather_daily.csv', parse_dates=['date'])
print(f"Date range: {df['date'].min()} to {df['date'].max()}")
print(f"Total days: {len(df)}")
df.head()
```

```python
# Target distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

df['temp_max_real'].hist(bins=50, ax=axes[0], color='steelblue')
axes[0].set_xlabel('Max Temperature (°C)')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Distribution of Daily Max Temperature')

# Monthly max temp
df['month'] = df['date'].dt.month
monthly = df.groupby('month')['temp_max_real'].agg(['mean', 'std', 'min', 'max'])
monthly.plot(kind='bar', y='mean', yerr='std', ax=axes[1], color='steelblue', capsize=3)
axes[1].set_xlabel('Month')
axes[1].set_ylabel('Max Temperature (°C)')
axes[1].set_title('Average Max Temp by Month (with std)')
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Correlation matrix
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr = df[numeric_cols].corr()

fig, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, ax=ax, fmt='.2f')
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
# Morning temp vs max temp
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(df['temp_06h'], df['temp_max_real'], alpha=0.3, s=5, color='steelblue')
ax.set_xlabel('Morning Temperature at 06h (°C)')
ax.set_ylabel('Daily Max Temperature (°C)')
ax.set_title('Morning Temp vs Max Temp')

# Add perfect prediction line
lims = [ax.get_xlim()[0], ax.get_xlim()[1]]
ax.plot(lims, lims, 'r--', alpha=0.5, label='Perfect prediction')
ax.legend()
plt.show()
```

```python
# Prediction error by month
df['temp_error'] = df['temp_max_real'] - df['temp_mean_real']

fig, ax = plt.subplots(figsize=(10, 4))
df.groupby('month')['temp_error'].mean().plot(kind='bar', ax=ax, color='steelblue')
ax.set_xlabel('Month')
ax.set_ylabel('Mean Error (°C)')
ax.set_title('Mean Difference: Max - Mean by Month')
ax.axhline(0, color='r', linestyle='--')
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
plt.tight_layout()
plt.show()
```
